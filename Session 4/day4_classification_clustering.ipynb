$z = X \cdot W + b$

، لكن النتيجة قد تكون أي رقم. لذا نمرّرها عبر <strong style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; color:#FFFFFF;">دالة السيغمويد (Sigmoid)</strong> التي تضغطها إلى احتمال بين 0 و 1:</p><div dir="ltr" style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; background:#0B0B0B; border-radius:8px; padding:8px 12px; text-align:center; color:#FFFFFF; margin:8px 0;">

$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

$$\hat{y} = \begin{cases} 1 & \text{if } \sigma(z) \ge 0.5 \\ 0 & \text{if } \sigma(z) < 0.5 \end{cases}$$

```python
# ==============================================================
# الخلية 1: الاستيراد ومنحنى دالة السيغمويد
# ==============================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.cluster import KMeans
from sklearn.metrics import (confusion_matrix, ConfusionMatrixDisplay, classification_report,
                             accuracy_score, precision_score, recall_score, f1_score,
                             precision_recall_curve, roc_auc_score, silhouette_score,
                             adjusted_rand_score)

# لوحة ألوان الهوية البصرية تُعرَّف مرة واحدة لتستخدمها كل الرسوم
RED, BLACK, GRAY = '#DC2626', '#0B0B0B', '#4B5563'
plt.rcParams.update({
    'figure.facecolor': 'white', 'axes.facecolor': 'white',
    'axes.grid': True, 'grid.alpha': 0.25,
    'axes.spines.top': False, 'axes.spines.right': False, 'font.size': 11,
})

# ---- منحنى السيغمويد ----
z = np.linspace(-8, 8, 400)
sigma = 1 / (1 + np.exp(-z))          # الدالة تحوّل أي رقم حقيقي إلى احتمال

plt.figure(figsize=(9, 4.8))
plt.plot(z, sigma, color=RED, linewidth=3, label=r'$\sigma(z) = 1/(1+e^{-z})$')
plt.axhline(0.5, color=BLACK, linestyle='--', linewidth=1.5, label='Decision threshold = 0.5')
plt.axvline(0, color=GRAY, linestyle=':', linewidth=1.2)
# المنطقة فوق العتبة تُصنَّف إيجابية، وما دونها سلبية
plt.fill_between(z, 0.5, sigma, where=sigma >= 0.5, color=RED, alpha=0.10)
plt.fill_between(z, sigma, 0.5, where=sigma < 0.5, color=BLACK, alpha=0.08)
plt.text(3, 0.2, 'Predict 0', color=BLACK, fontsize=12)
plt.text(1.5, 0.85, 'Predict 1', color=RED, fontsize=12)
plt.xlabel('z = X.W + b')
plt.ylabel('Probability')
plt.title('Sigmoid: from a linear score to a probability')
plt.legend(loc='upper left')
plt.show()
print('σ(0) =', sigma[np.argmin(np.abs(z))].round(3), '<- عند z = 0 الاحتمال 0.5 تماماً')
```

```python
# ==============================================================
# الخلية 2: بيانات غير متوازنة (1000 مريض) + تقسيم طبقي + تدريب
# ==============================================================
rng = np.random.default_rng(42)
n = 1000
# 15% حالات حرجة: نولّد الفئة أولاً ثم الميزات بتوزيعات متداخلة جزئياً (واقعية، لا فصل مثالي)
y_all = (rng.random(n) < 0.15).astype(int)
c = y_all == 1

def feat(mu0, sd0, mu1, sd1):
    """ميزة تختلف متوسطاتها بين السليم (0) والحرج (1)."""
    return np.where(c, rng.normal(mu1, sd1, n), rng.normal(mu0, sd0, n))

patients = pd.DataFrame({
    'Age': feat(55, 15, 62, 15).clip(18, 95).round(0),
    'Heart_Rate': feat(78, 12, 88, 16).clip(40, 180).round(0),
    'Systolic_BP': feat(125, 16, 115, 20).clip(60, 200).round(0),
    'Oxygen_Sat': feat(97, 2, 95, 3.5).clip(70, 100).round(1),
    'Lactate': feat(1.4, 0.7, 2.4, 1.3).clip(0.4, 10).round(2),      # مقياسه صغير جداً مقارنة بالضغط!
    'Temperature': feat(36.8, 0.5, 37.2, 0.8).round(1),
})
patients['Critical'] = y_all
print('نسبة الحالات الحرجة:', patients['Critical'].mean().round(3))
display(patients.head())

features = [col for col in patients.columns if col != 'Critical']
X, y = patients[features], patients['Critical']

# ---- لماذا stratify؟ نرى ما يحدث بدونها عبر عدة بذور ----
ratios = []
for seed in range(8):
    _, _, _, y_te = train_test_split(X, y, test_size=0.2, random_state=seed)   # بلا stratify
    ratios.append(y_te.mean())
print(f'\nنسبة الحرجة في الاختبار بلا stratify: من {min(ratios):.3f} إلى {max(ratios):.3f} (الأصل 0.150)')

# ---- التقسيم الصحيح ----
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)       # stratify يحافظ على نسبة 15% في الجزأين
print(f'مع stratify -> تدريب: {y_train.mean():.3f} | اختبار: {y_test.mean():.3f}')

# ---- التطبيع الآمن: fit على التدريب فقط ----
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)

# ---- التدريب ----
log_reg = LogisticRegression(random_state=42).fit(X_train_s, y_train)
# الغابة لا تحتاج تطبيعاً لأنها تقارن كل ميزة بعتبة، لذا نعطيها البيانات الأصلية
forest = RandomForestClassifier(n_estimators=200, random_state=42).fit(X_train, y_train)
print('\nتم تدريب LogisticRegression و RandomForestClassifier')
```

```python
# ==============================================================
# الخلية 3: المصنف الوهمي مقابل Logistic Regression و Random Forest
# ==============================================================
# مصنف وهمي يتوقع دائماً الفئة الأكثر تكراراً: هو "خط الأساس" الذي يجب على أي نموذج حقيقي أن يتفوق عليه
dummy = DummyClassifier(strategy='most_frequent').fit(X_train, y_train)

models = {
    'Dummy (always 0)': (dummy, X_test),
    'Logistic Regression': (log_reg, X_test_s),      # المطبَّعة
    'Random Forest': (forest, X_test),               # الأصلية
}

red_cmap = LinearSegmentedColormap.from_list('brand', ['#FFFFFF', RED])   # أبيض -> أحمر بهوية الدورة
rows = []
fig, axes = plt.subplots(1, 3, figsize=(14, 4.2))
for ax, (name, (model, X_eval)) in zip(axes, models.items()):
    pred = model.predict(X_eval)
    ConfusionMatrixDisplay.from_predictions(
        y_test, pred, ax=ax, cmap=red_cmap, colorbar=False,
        display_labels=['Healthy', 'Critical'])
    ax.grid(False)
    ax.set_title(name)
    rows.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, pred),
        'Precision': precision_score(y_test, pred, zero_division=0),   # zero_division يمنع التحذير عندما لا يتنبأ الوهمي بأي إيجابي
        'Recall': recall_score(y_test, pred),
        'F1': f1_score(y_test, pred, zero_division=0),
    })
plt.tight_layout()
plt.show()

summary = pd.DataFrame(rows).set_index('Model').round(3)
display(summary)
print('لاحظ: الوهمي يحقق دقة عالية جداً و Recall = 0! الدقة وحدها خدعة في البيانات غير المتوازنة.\n')

print('تقرير Logistic Regression:')
print(classification_report(y_test, log_reg.predict(X_test_s), target_names=['Healthy', 'Critical'], zero_division=0))
```

$\tau$

:</p><div dir="ltr" style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; background:#0B0B0B; border-radius:8px; padding:8px 12px; text-align:center; color:#FFFFFF; margin:8px 0;">

```python
# ==============================================================
# الخلية 4: تجربة العتبات الثلاث ومنحنى المقايضة
# ==============================================================
# [:, 1] هو احتمال الفئة الحرجة. العمود [:, 0] هو احتمال "سليم"
proba = log_reg.predict_proba(X_test_s)[:, 1]
thresholds = [0.25, 0.50, 0.75]

fig = plt.figure(figsize=(14, 8.5))
gs = fig.add_gridspec(2, 3, height_ratios=[1, 1.05])

rows = []
for i, tau in enumerate(thresholds):
    pred = (proba >= tau).astype(int)                 # نطبق عتبتنا نحن بدل 0.5 الصماء
    ax = fig.add_subplot(gs[0, i])
    ConfusionMatrixDisplay.from_predictions(y_test, pred, ax=ax, cmap=red_cmap, colorbar=False,
                                            display_labels=['Healthy', 'Critical'])
    ax.grid(False)
    ax.set_title(f'tau = {tau:.2f}')
    rows.append({'Threshold': tau, 'Precision': precision_score(y_test, pred, zero_division=0),
                 'Recall': recall_score(y_test, pred), 'FN': int(((pred == 0) & (y_test == 1)).sum()),
                 'FP': int(((pred == 1) & (y_test == 0)).sum())})

# منحنى المقايضة: كل نقطة عتبة ممكنة. الأحمر Recall والأسود Precision
prec, rec, thr = precision_recall_curve(y_test, proba)
ax = fig.add_subplot(gs[1, :])
ax.plot(thr, rec[:-1], color=RED, linewidth=2.5, label='Recall')
ax.plot(thr, prec[:-1], color=BLACK, linewidth=2.5, label='Precision')
for tau in thresholds:
    ax.axvline(tau, color=GRAY, linestyle='--', linewidth=1)
ax.set_xlabel('Decision threshold tau')
ax.set_ylabel('Score')
ax.set_title('Precision-Recall trade-off')
ax.legend()
plt.tight_layout()
plt.show()

display(pd.DataFrame(rows).round(3))

# ---- الفخ: العمود الخطأ ----
wrong = log_reg.predict_proba(X_test_s)[:, 0]
print('AUC باستخدام [:, 1] =', round(roc_auc_score(y_test, proba), 3))
print('AUC باستخدام [:, 0] =', round(roc_auc_score(y_test, wrong), 3), '<- أقل من 0.5 = ترتيب معكوس!')
```

max-width:100%; overflow-wrap:anywhere; color:#FCA5A5;">الكارثة الشائعة: تشغيل KMeans بدون تطبيع البيانات (StandardScaler):</strong> لأن K-Means تحسب المسافة الإقليدية، فإن عدم التطبيع يجعل الخوارزمية تقسم البيانات بناءً على العمود ذي الأرقام الأكبر فقط!</li><li style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; background:#0B0B0B; color:#FCA5A5; padding:1px 6px; border-radius:4px; font-family:Consolas,monospace; font-size:13px;">df.groupby('Cluster').mean()</code>) ليطلق عليها اسماً تجارياً ذا معنى (مثل: "عملاء VIP"، "عملاء مقتصدون").</li></ul></div>

```python
# ==============================================================
# الخلية 5: بيانات سلوك العملاء وإيجاد K الأمثل
# ==============================================================
rng_c = np.random.default_rng(7)
# أربع شرائح حقيقية بشكل شبكة 2x2: (إنفاق منخفض/مرتفع) × (تكرار منخفض/مرتفع)
# نعرفها نحن فقط لأننا من يولّد البيانات؛ الخوارزمية لن ترى هذه التسميات
seg_params = [(2000, 500, 6, 2), (2200, 500, 34, 4), (9000, 1100, 8, 2.5), (9500, 1100, 36, 4)]
parts, true_seg = [], []
for s, (m_sp, sd_sp, m_fr, sd_fr) in enumerate(seg_params):
    k_n = 150
    parts.append(np.column_stack([rng_c.normal(m_sp, sd_sp, k_n), rng_c.normal(m_fr, sd_fr, k_n)]))
    true_seg += [s] * k_n
customers = pd.DataFrame(np.vstack(parts), columns=['Annual_Spend', 'Purchase_Frequency'])
customers['Annual_Spend'] = customers['Annual_Spend'].clip(100).round(0)
customers['Purchase_Frequency'] = customers['Purchase_Frequency'].clip(1).round(0)
true_seg = np.array(true_seg)
display(customers.describe().round(1))
# الإنفاق بالآلاف والتكرار بالعشرات: فرق مقياس كبير يجعل التطبيع إلزامياً

# التطبيع: كل ميزة بمتوسط 0 وانحراف 1 كي تتساوى في وزن المسافة
cust_scaler = StandardScaler()
C = cust_scaler.fit_transform(customers[['Annual_Spend', 'Purchase_Frequency']])

ks = range(2, 11)
inertias, sils = [], []
for k in ks:
    # n_init=10: نكرر التجربة بمراكز بدء مختلفة ونأخذ الأفضل، لأن البداية العشوائية السيئة تعطي نتيجة سيئة
    km_k = KMeans(n_clusters=k, n_init=10, random_state=42).fit(C)
    inertias.append(km_k.inertia_)
    sils.append(silhouette_score(C, km_k.labels_))

best_k = list(ks)[int(np.argmax(sils))]

fig, axes = plt.subplots(1, 2, figsize=(13, 4.6))
axes[0].plot(list(ks), inertias, color=RED, marker='o', linewidth=2.5)
axes[0].set_title('Elbow curve (Inertia)')
axes[0].set_xlabel('K'); axes[0].set_ylabel('Inertia')
axes[1].plot(list(ks), sils, color=BLACK, marker='o', linewidth=2.5)
axes[1].axvline(best_k, color=RED, linestyle='--', label=f'Best K = {best_k}')
axes[1].set_title('Silhouette score'); axes[1].set_xlabel('K'); axes[1].legend()
plt.tight_layout()
plt.show()
print('أفضل K بحسب Silhouette:', best_k)

# ---- فخ: ماذا يحدث بلا تطبيع؟ نقارن اتفاق التجميع مع الشرائح الحقيقية (ARI=1 تطابق تام) ----
raw_labels = KMeans(n_clusters=4, n_init=10, random_state=42).fit_predict(customers[['Annual_Spend', 'Purchase_Frequency']])
scaled_labels = KMeans(n_clusters=4, n_init=10, random_state=42).fit_predict(C)
print('\nاتفاق التجميع مع الشرائح الحقيقية (ARI):')
print('  بلا تطبيع :', round(adjusted_rand_score(true_seg, raw_labels), 3))
print('  مع تطبيع  :', round(adjusted_rand_score(true_seg, scaled_labels), 3))
```

```python
# ==============================================================
# الخلية 6: رسم المجموعات وتسميتها تجارياً
# ==============================================================
final_km = KMeans(n_clusters=best_k, n_init=10, random_state=42).fit(C)
customers['Cluster'] = final_km.labels_

# المراكز محسوبة في الفضاء المطبَّع؛ نعيدها للوحدات الأصلية لتصبح مفهومة (دولار ومرات شراء)
centers = cust_scaler.inverse_transform(final_km.cluster_centers_)

shades = ['#0B0B0B', '#4B5563', '#9CA3AF', '#FCA5A5', '#374151', '#D1D5DB']   # تدرجات الهوية للمجموعات
plt.figure(figsize=(9, 6))
for k in range(best_k):
    pts = customers[customers['Cluster'] == k]
    plt.scatter(pts['Annual_Spend'], pts['Purchase_Frequency'], s=26, color=shades[k % len(shades)], alpha=0.75, label=f'Cluster {k}')
plt.scatter(centers[:, 0], centers[:, 1], marker='X', s=320, color=RED, edgecolor='white', linewidth=1.5, label='Centroids', zorder=5)
plt.xlabel('Annual spend (USD)')
plt.ylabel('Purchase frequency (per year)')
plt.title(f'K-Means customer segments (K = {best_k})')
plt.legend()
plt.show()

# ---- التحليل التجاري: الأرقام وحدها بلا معنى حتى نقرأ المتوسطات ----
profile = customers.groupby('Cluster').agg(
    Customers=('Annual_Spend', 'size'),
    Avg_Spend=('Annual_Spend', 'mean'),
    Avg_Frequency=('Purchase_Frequency', 'mean'),
).round(1)

# نمنح كل مجموعة اسماً تجارياً بمقارنة متوسطيها بالوسيط العام: رقم المجموعة نفسه عشوائي ولا يدل على شيء
def business_name(row):
    high_spend = row['Avg_Spend'] > customers['Annual_Spend'].median()
    high_freq = row['Avg_Frequency'] > customers['Purchase_Frequency'].median()
    if high_spend and high_freq:
        return 'VIP customers'
    if high_spend:
        return 'Big-ticket occasional'
    if high_freq:
        return 'Frequent small-basket'
    return 'Budget occasional'

profile['Segment'] = profile.apply(business_name, axis=1)
display(profile.sort_values(['Avg_Spend', 'Avg_Frequency']))
```
