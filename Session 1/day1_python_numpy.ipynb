```python
# ==============================================================
# المحور A2: الدوال المتقدمة
# ==============================================================
readings = [("s1", 21.5), ("s2", 35.2), ("s3", -999.0), ("s4", 28.1)]   # -999 = عطل في الحساس

# filter: نبقي القراءات السليمة فقط
valid = list(filter(lambda r: r[1] > -100, readings))
# map: نحوّل من مئوي إلى فهرنهايت
fahrenheit = list(map(lambda r: round(r[1] * 9 / 5 + 32, 1), valid))
# sorted + key: الترتيب بالقيمة (العنصر الثاني) من الأعلى للأدنى
ranked = sorted(valid, key=lambda r: r[1], reverse=True)
print("السليمة :", valid)
print("فهرنهايت:", fahrenheit)
print("الأعلى حرارة:", ranked[0])

# *args تجمع القيم الموضعية، و**options تجمع المسماة: دالة مرنة لأي عدد من القراءات
def summarize(*values, **options):
    digits = options.get("digits", 2)           # قيمة افتراضية إن لم تُمرَّر
    return {"count": len(values), "mean": round(sum(values) / len(values), digits), "max": max(values)}

temps = [r[1] for r in valid]
print("\nالملخص:", summarize(*temps, digits=1))   # النجمة تفكّ القائمة إلى معاملات منفصلة

# ---- الفخ 1: map ترجع iterator ----
m = map(lambda t: t * 2, temps)
print("\nطباعة map مباشرة:", str(m)[:22], "...")
print("بعد list():", list(m))

# ---- الفخ 2: الربط المتأخر ----
fs = [lambda: i for i in range(3)]
print("\nالخاطئ :", [f() for f in fs])
fs = [lambda i=i: i for i in range(3)]       # القيمة الافتراضية تثبّت i لحظة التعريف
print("الصحيح :", [f() for f in fs])
```

```python
# ==============================================================
# المحور A3: القواميس والمجموعات
# ==============================================================
# سجلات نتائج نموذج تصنيف صور (كما تصل من API)
records = [
    {"id": 1, "label": "cat",  "conf": 0.93},
    {"id": 2, "label": "dog",  "conf": 0.88},
    {"id": 3, "label": "cat",  "conf": 0.71},
    {"id": 4, "label": "bird", "conf": 0.95},
    {"id": 5, "label": "dog",  "conf": 0.64},
    {"id": 6, "label": "cat",  "conf": 0.82},
]

# Counter: توزيع الفئات (أول خطوة في أي مشروع تصنيف: هل الفئات متوازنة؟)
counts = Counter(r["label"] for r in records)
print("التوزيع:", counts, "| الأكثر تكراراً:", counts.most_common(1))

# defaultdict(list): نجمّع الثقة لكل فئة دون فحص "هل المفتاح موجود؟"
by_label = defaultdict(list)
for r in records:
    by_label[r["label"]].append(r["conf"])
avg_conf = {label: round(sum(v) / len(v), 3) for label, v in by_label.items()}
print("متوسط الثقة لكل فئة:", avg_conf)

# المجموعات: كشف تسرب البيانات (Data Leakage) بين التدريب والاختبار
train_ids = {101, 102, 103, 104, 105}
test_ids = {104, 106, 107}
leak = train_ids & test_ids          # التقاطع = عينات ظهرت في الجانبين!
print("\nعينات متسربة:", leak)

# ---- الفخ 1: KeyError ----
config = {"lr": 0.01}
try:
    config["epochs"]
except KeyError as e:
    print("\nKeyError:", e)
print("مع get:", config.get("epochs", 10))

# ---- الفخ 2: التعديل أثناء المرور ----
d = {"a": 1, "b": -2, "c": 3, "d": -4}
try:
    for k in d:
        if d[k] < 0:
            del d[k]
except RuntimeError as e:
    print("\nRuntimeError:", e)
for k in list(d):                    # list(d) نسخة مستقلة من المفاتيح
    if d[k] < 0:
        del d[k]
print("بعد الحذف الآمن:", d)
```

```python
# ==============================================================
# تمرين 2: إصلاح أعطال الحساسات بقناع منطقي
# ==============================================================
sensor = np.array([22.1, -999.0, 24.3, 23.8, -999.0, 25.0, 21.7])

# TODO (جرّب بنفسك أولاً)
# حل مقترح:
bad = sensor == -999.0                    # قناع الأعطال
mean_ok = sensor[~bad].mean()             # المتوسط من السليمة فقط؛ لو حسبناه مع -999 لتلوّث
fixed = sensor.copy()                     # نعمل على نسخة كي تبقى البيانات الأصلية للتوثيق
fixed[bad] = mean_ok
print("المتوسط السليم:", round(mean_ok, 2), "| الأعطال المُصلحة:", int(bad.sum()))
print(fixed.round(2))

assert bad.sum() == 2 and fixed.min() > 0
assert abs(fixed.mean() - mean_ok) < 1e-9
```

```python
# ==============================================================
# المحور B4: البث والتجميع
# ==============================================================
# 5 أشخاص × 3 ميزات: العمر، الطول (سم)، الوزن (كغ). كل ميزة بمقياس مختلف جداً!
X = rng.normal(loc=[50, 170, 70], scale=[10, 8, 12], size=(5, 3)).round(1)
print("X =\n", X)
print("\nمتوسط كل ميزة (axis=0):", X.mean(axis=0).round(1), "| shape", X.mean(axis=0).shape)
print("متوسط كل عينة (axis=1):", X.mean(axis=1).round(1), "| shape", X.mean(axis=1).shape)

# ---- التطبيع بالبث: (5,3) - (3,) يعمل لأن المتجه يُمدَّد على الصفوف ----
Z = (X - X.mean(axis=0)) / X.std(axis=0)
print("\nمتوسط الميزات بعد التطبيع:", Z.mean(axis=0).round(8))   # ≈ 0
print("انحراف الميزات بعد التطبيع:", Z.std(axis=0).round(8))      # = 1

# ---- keepdims: تحويل كل صف إلى توزيع احتمالي (مجموعه 1) ----
P = np.abs(X)
probs = P / P.sum(axis=1, keepdims=True)    # keepdims يبقي الشكل (5,1) فيتم البث بشكل صحيح على الأعمدة
print("مجموع كل صف:", probs.sum(axis=1).round(6))

# ---- الفخ 1: أشكال غير متوافقة ----
A = np.ones((3, 4))
v = np.array([1, 2, 3])
try:
    A + v
except ValueError as e:
    print("\nValueError:", e)
print("الحل v[:, None] -> الشكل:", (A + v[:, None]).shape)

# ---- الفخ 2: NaN ----
vals = np.array([1.0, np.nan, 3.0])
print("\nmean   :", vals.mean(), "| nanmean:", np.nanmean(vals))
```

$$\text{softmax}(z)_i = \frac{e^{z_i - \max(z)}}{\sum_j e^{z_j - \max(z)}}$$

```python
# ==============================================================
# كود البداية للتحدي (Starter Code)
# ==============================================================
rng_c = np.random.default_rng(7)
images = rng_c.integers(0, 256, size=(8, 8, 8), dtype=np.uint8)       # 8 صور، كل منها 8x8
W = rng_c.normal(0, 0.1, size=(64, 10))                                # أوزان الطبقة
b = np.zeros(10)                                                       # انحياز كل فئة
print("images:", images.shape, "| W:", W.shape, "| b:", b.shape)

# TODO 1: سطّح images إلى (8, 64) وحوّلها إلى float مقسومة على 255
# TODO 2: logits = X @ W + b   (الشكل المتوقع: (8, 10))
# TODO 3: softmax مستقر: اطرح الأكبر في كل صف قبل exp، واقسم على المجموع (axis=1, keepdims=True)
# TODO 4: predictions = argmax على axis=1  (الشكل المتوقع: (8,))
```

```python
# ==============================================================
# الحل النموذجي + التحقق الآلي
# ==============================================================
# 1) تسطيح وتطبيع: float لأن القسمة على 255 تعطي كسوراً، و uint8 سيفسدها
X = images.reshape(8, -1).astype(np.float64) / 255.0

# 2) الطبقة الخطية: سطر واحد لكل الصور
logits = X @ W + b

# 3) لماذا نطرح الأكبر؟ لأن exp للأرقام الكبيرة يفيض إلى inf فتصبح النتيجة nan
big = np.array([[1000.0, 1001.0, 1002.0]])
with np.errstate(over="ignore", invalid="ignore"):
    naive = np.exp(big) / np.exp(big).sum(axis=1, keepdims=True)
print("softmax الساذج على قيم كبيرة:", naive, "<- nan!")

def stable_softmax(z):
    z = z - z.max(axis=1, keepdims=True)             # أكبر قيمة تصبح 0، فيكون exp <= 1 دائماً
    e = np.exp(z)
    return e / e.sum(axis=1, keepdims=True)          # keepdims يضمن بثاً صحيحاً على الأعمدة

print("softmax المستقر على نفس القيم :", stable_softmax(big).round(3))

probs = stable_softmax(logits)

# 4) الفئة المتوقعة لكل صورة
predictions = probs.argmax(axis=1)
print("\nالتوقعات:", predictions)
print("احتمال أعلى فئة للصورة الأولى:", round(float(probs[0].max()), 3))

# ---------------- الاختبارات الآلية ----------------
assert X.shape == (8, 64) and X.min() >= 0 and X.max() <= 1, "التطبيع خاطئ"
assert logits.shape == (8, 10)
assert probs.shape == (8, 10) and np.all(probs >= 0)
assert np.allclose(probs.sum(axis=1), 1.0), "كل صف يجب أن يجمع 1"
assert predictions.shape == (8,) and np.array_equal(predictions, logits.argmax(axis=1))
assert not np.isnan(stable_softmax(big)).any()

print("\033[92m✔ نجحت جميع الاختبارات! بنيتَ طبقة عصبية كاملة بـ NumPy فقط.\033[0m")
```
