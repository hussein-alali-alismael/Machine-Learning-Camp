```python
# ==============================================================
# المحور 2: العوامل
# ==============================================================
import math

a, b = 7, 2
print("a / b  =", a / b, "   (قسمة كاملة: الناتج float دائماً)")
print("a // b =", a // b, "     (قسمة صحيحة: تُسقط الكسر)")
print("a % b  =", a % b, "     (الباقي: مفيد لفحص الزوجي/الفردي)")
print("2 ** 3 =", 2 ** 3, "    (القوة)")

# الإسناد المركب: اختصار يوفر التكرار ويقلل أخطاء الكتابة
counter = 10
counter += 5
counter *= 2
print("\ncounter =", counter)

# المنطق والعضوية
age, has_id = 20, True
print("\nيدخل؟", age >= 18 and has_id)
print("'th' in 'python' ->", "th" in "python")

# ---- الفخ 1: الإسناد داخل الشرط ----
try:
    compile("if age = 5: pass", "<demo>", "exec")   # نختبر الصياغة فقط دون تنفيذ
except SyntaxError as e:
    print("\nSyntaxError:", e.msg)

# ---- الفخ 2: ^ ليست رفعاً للقوة ----
print("\n2 ^ 3  =", 2 ^ 3, " <- XOR وليس الأسس!")
print("2 ** 3 =", 2 ** 3, " <- الأسس")

# ---- الفخ 3: دقة الأعداد العشرية ----
print("\n0.1 + 0.2 =", 0.1 + 0.2)
print("0.1 + 0.2 == 0.3                  :", 0.1 + 0.2 == 0.3)
print("round(0.1 + 0.2, 2) == 0.3        :", round(0.1 + 0.2, 2) == 0.3)
# isclose أفضل هندسياً لأنها تقارن ضمن هامش خطأ صغير بدل تقريب يدوي
print("math.isclose(0.1 + 0.2, 0.3)      :", math.isclose(0.1 + 0.2, 0.3))
```

```python
# ==============================================================
# المحور 4: القوائم والصفوف
# ==============================================================
# ---- فخ المرجع: id() يثبت أن الاسمين يشيران لنفس الكائن ----
list_a = [1, 2, 3]
list_b = list_a                    # ليست نسخة!
list_b.append(99)
print("list_a =", list_a, "<- تغيّرت دون أن نلمسها!")
print("نفس الكائن؟", id(list_a) == id(list_b))

list_c = list_a.copy()             # نسخة حقيقية في مكان مختلف من الذاكرة
list_c.append(100)
print("بعد copy: list_a =", list_a, "| list_c =", list_c, "| نفس الكائن؟", id(list_a) == id(list_c))

# ---- sort() مقابل sorted() ----
nums = [3, 1, 2]
result = nums.sort()
print("\nsort() ترجع:", result, "| nums =", nums)       # النتيجة None والقائمة رُتّبت في مكانها
nums = [3, 1, 2]
new_sorted = sorted(nums)                                 # sorted لا تمس الأصل وتُرجع قائمة جديدة
print("sorted() ترجع:", new_sorted, "| الأصل لم يتغير:", nums)

# ---- append مقابل extend ----
a = [1, 2]; a.append([3, 4])
b = [1, 2]; b.extend([3, 4])
print("\nappend:", a, "| extend:", b)

# ---- فخ الصف ذي العنصر الواحد ----
print("\ntype((5))  =", type((5)).__name__)
print("type((5,)) =", type((5,)).__name__)

# ---- الصفوف: عدم التعديل والتفكيك ----
point = (3, 4)
try:
    point[0] = 10
except TypeError as e:
    print("\nTypeError:", e)

x, y = point                       # تفكيك: كل قيمة تذهب لمتغير
print("x =", x, "| y =", y)
first, *rest = [10, 20, 30, 40]    # النجمة تلتقط الباقي في قائمة
print("first =", first, "| rest =", rest)
x, y = y, x                        # تبديل قيمتين بسطر واحد بفضل التفكيك
print("بعد التبديل: x =", x, "| y =", y)
```

```python
# ==============================================================
# تمرين 2: تنظيف الدرجات واستخراج أعلى 3
# ==============================================================
grades = [88, "92", "abc", 75, " 60 ", None, 95.5, "100", -3, 240]

# TODO (جرّب بنفسك أولاً)
# حل مقترح:
clean = []
for g in grades:
    try:
        value = float(str(g).strip())      # نحوّل عبر str أولاً ليتحمل الكود الأنواع المختلفة
    except ValueError:
        continue                           # نتجاهل ما لا يُقرأ كرقم (abc, None)
    if 0 <= value <= 100:                  # نستبعد القيم المستحيلة (-3, 240)
        clean.append(value)

top3 = sorted(clean, reverse=True)[:3]     # sorted تعطي قائمة جديدة فلا نمس clean
print("الدرجات النظيفة:", clean)
print("أعلى 3:", top3)
print("القائمة الأصلية سليمة:", grades)

assert top3 == [100.0, 95.5, 92.0]
assert len(grades) == 10 and grades[2] == "abc"
```

```python
# ==============================================================
# المحور 5: الجمل الشرطية
# ==============================================================
score = 95

# ---- النسخة الصحيحة: if / elif / else ----
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"
print("elif  ->", grade)

# ---- الفخ 1: سلسلة if متتالية ----
grade_bad = None
if score >= 90:
    grade_bad = "A"
if score >= 80:          # يُفحص رغم أن الشرط الأول تحقق!
    grade_bad = "B"
if score >= 70:
    grade_bad = "C"
else:
    grade_bad = "F"      # هذه else مرتبطة بالـ if الأخيرة فقط
print("if متتالية ->", grade_bad, " <- نتيجة خاطئة لدرجة 95!")

# ---- الفخ 2: if x == 1 or 2 ----
x = 999
if x == 1 or 2:
    print("\nالشرط تحقق رغم أن x = 999 !")
print("bool(2) =", bool(2), "<- هذا سبب المشكلة")
print("الحل 1 :", x == 1 or x == 2)
print("الحل 2 :", x in (1, 2))

# ---- Truthy / Falsy ----
print("\nالقيم الفارغة:")
for value in [0, 0.0, "", [], None, "0", [0]]:
    print(f"  {value!r:6} -> {bool(value)}")

my_list = []
if not my_list:          # طريقة أنيقة وقياسية لفحص القائمة الفارغة
    print("\nالقائمة فارغة")

# ---- Ternary ----
status = "Pass" if score >= 60 else "Fail"
print("status =", status)
```

```python
# ==============================================================
# المحور 6: الحلقات
# ==============================================================
print("range(1, 5)     ->", list(range(1, 5)), "<- تتوقف عند 4")
print("range(1, 10, 3) ->", list(range(1, 10, 3)))

# ---- for مع break و continue ----
for n in range(1, 10):
    if n % 2 == 0:
        continue           # نتخطى الأعداد الزوجية فقط
    if n > 7:
        break              # نكسر الحلقة كلها عند تجاوز 7
    print("فردي:", n)

# ---- while مع حارس أمان ----
count = 0
while count < 3:
    print("count =", count)
    count += 1             # السطر الذي يقرّب الشرط من الفشل؛ بدونه تصبح الحلقة لانهائية

# ---- enumerate و zip ----
names = ["Sara", "Omar", "Lina"]
marks = [90, 72, 85]
for i, name in enumerate(names, start=1):   # start=1 ليبدأ الترقيم من 1 كما يتوقع البشر
    print(i, name)
for name, mark in zip(names, marks):         # zip تقرن العنصر i من القائمتين معاً
    print(f"{name}: {mark}")

# ---- الفخ: الحذف أثناء المرور ----
nums = [-1, -2, 3, -4, 5, -6]
broken = nums.copy()
for x in broken:
    if x < 0:
        broken.remove(x)   # الحذف يزحزح العناصر فيقفز المؤشر فوق التالي
print("\nالحذف المباشر ->", broken, "<- ما زال فيه سالب (-2)!")

# الحل: نبني قائمة جديدة بدل تعديل القائمة أثناء المرور عليها
positives = []
for x in nums:
    if x >= 0:
        positives.append(x)
print("الحل الصحيح   ->", positives)

fixed = nums.copy()
for x in fixed.copy():     # حل بديل: نمرّ على نسخة ونعدّل الأصل
    if x < 0:
        fixed.remove(x)
print("الحل بالنسخة  ->", fixed)
```

```python
# ==============================================================
# تمرين 3: لوحة الشرف
# ==============================================================
names = ["Omar", "Sara", "Lina", "Zaid"]
scores = [72, 90, 45, 85]

# TODO (جرّب بنفسك أولاً)
# حل مقترح:
pairs = []
for name, score in zip(names, scores):
    if score >= 50:
        pairs.append((score, name))        # نضع الدرجة أولاً لأن sorted ترتب حسب أول عنصر

leaderboard = []
for rank, (score, name) in enumerate(sorted(pairs, reverse=True), start=1):
    leaderboard.append(f"{rank}. {name}: {score}")

print("\n".join(leaderboard))
assert leaderboard == ["1. Sara: 90", "2. Zaid: 85", "3. Omar: 72"]
```

```python
# ==============================================================
# المحور 7: الدوال
# ==============================================================
def calc_bmi(weight, height=1.7):
    """تُرجع مؤشر كتلة الجسم. الطول بالمتر ولديه قيمة افتراضية."""
    return round(weight / height ** 2, 1)

print("calc_bmi(70, 1.75) =", calc_bmi(70, 1.75))
print("calc_bmi(70)       =", calc_bmi(70), "<- استخدمت الطول الافتراضي")

# ---- إرجاع عدة قيم (tuple) وتفكيكها ----
def analyze(weight, height):
    bmi = calc_bmi(weight, height)
    label = "ideal" if 18.5 <= bmi <= 24.9 else "not ideal"
    return bmi, label                  # نرجع صفاً: قيمتان معاً

bmi_value, bmi_label = analyze(70, 1.75)
print("\nالتفكيك:", bmi_value, bmi_label)

# ---- الفخ 1: print بدل return ----
def double_print(n):
    print(n * 2)                       # تعرض على الشاشة فقط، لا تُرجع شيئاً

def double_return(n):
    return n * 2

val = double_print(5)
print("val =", val, "<- None!")
try:
    val + 1
except TypeError as e:
    print("TypeError:", e)
print("مع return:", double_return(5) + 1)

# ---- الفخ 2: معامل افتراضي قبل عادي ----
try:
    compile("def calc(a=10, b): pass", "<demo>", "exec")
except SyntaxError as e:
    print("\nSyntaxError:", e.msg)

# ---- الفخ 3: القائمة كقيمة افتراضية ----
def add_student_bad(name, group=[]):
    group.append(name)
    return group

print("\nالنسخة الخاطئة:")
print(add_student_bad("Ali"))
print(add_student_bad("Sara"), "<- Ali ما زال موجوداً من الاستدعاء السابق!")

def add_student(name, group=None):
    if group is None:                  # نصنع قائمة جديدة عند كل استدعاء
        group = []
    group.append(name)
    return group

print("\nالنسخة الصحيحة:")
print(add_student("Ali"))
print(add_student("Sara"))

# ---- النطاق (Scope) ----
total = 100                            # Global
def show_scope():
    local_var = 5                      # Local
    return total + local_var           # القراءة من Global مسموحة
print("\nshow_scope() =", show_scope())
print("'local_var' معرّف خارج الدالة؟", "local_var" in globals())
```

```python
# ==============================================================
# الحلول النموذجية + الاختبار الآلي
# ==============================================================

def parse_sensor_logs(logs):
    """تُرجع قائمة بالقراءات العشرية الصالحة فقط."""
    valid = []
    for log in logs:
        text = log.strip().lower()             # نوحّد الشكل أولاً حتى لا نفرّق بين TEMP وtemp
        if not text.startswith("temp:"):
            continue                           # نتجاهل ERROR وغيرها
        number_part = text.split(":")[1].replace("c", "").strip()
        try:
            value = float(number_part)
        except ValueError:
            continue                           # قراءة تالفة مثل "temp: abc c"
        if value < 0:
            continue                           # القيم السالبة مرفوضة بحسب المطلوب
        valid.append(value)
    return valid


def analyze_scores(scores, pass_mark=60):
    """تُرجع (المتوسط، الدرجات الناجحة، عدد الراسبين)."""
    if not scores:                             # نحمي الدالة من القسمة على صفر
        return 0.0, [], 0
    passed = []
    failed_count = 0
    for s in scores:
        if s >= pass_mark:
            passed.append(s)
        else:
            failed_count += 1
    return sum(scores) / len(scores), passed, failed_count


def merge_inventory(names, quantities):
    """تدمج القائمتين وتستبعد المنتجات منتهية الكمية."""
    inventory = []
    for name, qty in zip(names, quantities):
        if qty > 0:
            inventory.append((name, qty))
    return inventory


# ---------------- الاختبارات ----------------
logs = ["TEMP: 24.5 C", "ERROR", "temp: 31.0 c", "TEMP: -5.0 C", "TEMP: abc C"]
assert parse_sensor_logs(logs) == [24.5, 31.0]
assert parse_sensor_logs([]) == []

avg, passed, failed = analyze_scores([50, 60, 70, 90])
assert avg == 67.5 and passed == [60, 70, 90] and failed == 1
assert analyze_scores([]) == (0.0, [], 0)
assert analyze_scores([40, 55], pass_mark=50)[1] == [55]

assert merge_inventory(["pen", "book", "bag"], [10, 0, 3]) == [("pen", 10), ("bag", 3)]
assert merge_inventory(["a"], [-1]) == []

print("\033[92m✔ نجحت جميع الاختبارات! أحسنت، أساسياتك في بايثون صلبة.\033[0m")
```
