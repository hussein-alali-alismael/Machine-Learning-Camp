```python
# ==============================================================
# الخلية 1: توليد مجموعة بيانات طبية "فوضوية" تحاكي الواقع
# ==============================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# لماذا نثبّت البذرة؟ لكي يحصل كل طالب على نفس الأرقام تماماً،
# فتتطابق نتائجه مع الشرح على الشاشة ويسهل اكتشاف الأخطاء.
rng = np.random.default_rng(42)

N_UNIQUE = 485   # مرضى فريدون
N_DUPES = 15     # صفوف سنكررها عمداً -> المجموع 500 سجل

# ---------- 1) نولّد القيم "النظيفة" أولاً ----------
# لماذا؟ لأن الفوضى الواقعية تُضاف فوق حقيقة موجودة؛ هكذا نعرف ما الذي أفسدناه.
patient_id = np.arange(1001, 1001 + N_UNIQUE)
age_true = rng.normal(52, 13, N_UNIQUE).clip(21, 88).round()
height = rng.normal(168, 9, N_UNIQUE).clip(145, 198).round()
bmi_true = rng.normal(27, 4.5, N_UNIQUE).clip(17, 45)
# نشتق الوزن من BMI والطول حتى تبقى العلاقة الفيزيائية بينهما منطقية
weight = (bmi_true * (height / 100) ** 2).round(1)
smoking = rng.choice(['Never', 'Former', 'Current'], N_UNIQUE, p=[0.5, 0.25, 0.25])
smoke_code = pd.Series(smoking).map({'Never': 0, 'Former': 1, 'Current': 2}).values

# ضغط الدم والكوليسترول يرتفعان مع العمر والتدخين: هذا ما سيجعل التعويض "حسب الفئة" أذكى لاحقاً
systolic = (105 + 0.45 * (age_true - 20) + 4 * smoke_code + rng.normal(0, 9, N_UNIQUE)).round()
chol_true = (165 + 0.8 * (age_true - 20) + 15 * smoke_code + rng.normal(0, 22, N_UNIQUE)).round()
pulse = rng.normal(74, 9, N_UNIQUE).clip(45, 120).round()
city_clean = rng.choice(['Aleppo', 'Damascus', 'Homs', 'Hama'], N_UNIQUE, p=[0.35, 0.30, 0.20, 0.15])

# الهدف Heart_Risk يُولَّد من القيم النظيفة (قبل الإفساد) عبر دالة لوجستية
logit = (-0.9 + 0.045 * (age_true - 52) + 0.03 * (systolic - 125)
         + 0.012 * (chol_true - 200) + 0.05 * (bmi_true - 27) + 0.6 * (smoke_code - 0.75))
p_risk = 1 / (1 + np.exp(-logit))
heart_risk = (rng.random(N_UNIQUE) < p_risk).astype(int)

# ---------- 2) نُفسد البيانات عمداً ----------
# (أ) العمر: 6% قيم مفقودة + قيم مستحيلة + قيم متطرفة ممكنة لكنها شاذة
age = age_true.astype(float)
perm = rng.permutation(N_UNIQUE)
age[perm[:29]] = np.nan                        # فراغات
age[perm[29:33]] = [-5, 190, -1, 200]          # قيم مستحيلة طبياً
age[perm[33:36]] = [105, 112, 109]             # شواذ إحصائية (ممكنة لكنها بعيدة جداً)

# (ب) ضغط الدم: نص بمسافات ووحدات + كلمة Unknown + شواذ
bp = []
for v in systolic:
    r = rng.random()
    if r < 0.06:
        bp.append('Unknown')
    elif r < 0.30:
        bp.append(f' {int(v)} mmHg')
    elif r < 0.36:
        bp.append(f'{int(v)} mmHg ')
    else:
        bp.append(str(int(v)))
for i, val in zip(rng.choice(N_UNIQUE, 5, replace=False), [' 310 mmHg', '300', '280 mmHg', '25', ' 30 mmHg']):
    bp[i] = val

# (ج) الكوليسترول: 12% فراغات + قيم متطرفة جداً
chol = chol_true.astype(float)
perm2 = rng.permutation(N_UNIQUE)
chol[perm2[:58]] = np.nan
chol[perm2[58:62]] = [820, 950, 880, 35]

# (د) المدينة: عدم توحيد في الحروف والمسافات
def messy_city(c):
    r = rng.random()
    if r < 0.55: return c
    if r < 0.70: return c.lower() + ' '
    if r < 0.82: return c.upper()
    return ' ' + c + ' '
city = [messy_city(c) for c in city_clean]

# (هـ) حالة التدخين: 4% فراغات
smoking_col = pd.Series(smoking, dtype=object)
smoking_col.iloc[rng.choice(N_UNIQUE, 19, replace=False)] = np.nan

df_raw = pd.DataFrame({
    'Patient_ID': patient_id, 'Age': age, 'Blood_Pressure': bp, 'Cholesterol': chol,
    'City': city, 'Smoking_Status': smoking_col, 'Weight_kg': weight,
    'Height_cm': height, 'Pulse': pulse, 'Heart_Risk': heart_risk,
})

# ---------- 3) نضيف 15 سجلاً مكرراً ونخلط الترتيب ----------
# لماذا نخلط؟ في الواقع المكررات لا تأتي متتالية، ولا نريد للطالب أن يعتمد على ترتيب الصفوف.
dupes = df_raw.sample(N_DUPES, random_state=7)
df_raw = pd.concat([df_raw, dupes]).sample(frac=1, random_state=1).reset_index(drop=True)

# نحفظ ثم نقرأ من الملف، تماماً كما سيحدث مع بيانات حقيقية قادمة من مستشفى
df_raw.to_csv('dirty_medical_records.csv', index=False)
df = pd.read_csv('dirty_medical_records.csv')

print('تم إنشاء الملف dirty_medical_records.csv')
print('الأبعاد (صفوف, أعمدة):', df.shape)
```

```python
# ==============================================================
# الخلية 4: إزالة المكررات + توحيد النصوص + إصلاح الأنواع
# ==============================================================

# ---- المرحلة 1: المكررات ----
n_before = len(df)
# نعيد ترقيم الفهرس بعد الحذف حتى لا تبقى ثغرات في Index تربكنا لاحقاً
df = df.drop_duplicates().reset_index(drop=True)
print(f'حُذف {n_before - len(df)} صفاً مكرراً (الصفوف الآن: {len(df)})')
assert len(df) == 485

# ---- المرحلة 2: توحيد النصوص ----
print('\nقيم المدينة قبل التوحيد:', sorted(df['City'].unique()))
for col in ['City', 'Smoking_Status']:
    # strip يزيل المسافات الزائدة، وtitle يوحّد الحروف (aleppo -> Aleppo)
    # الفراغات NaN تبقى كما هي ولا تسبب خطأ
    df[col] = df[col].str.strip().str.title()
print('قيم المدينة بعد التوحيد :', sorted(df['City'].unique()))

# ---- المرحلة 3: إصلاح ضغط الدم ----
print('\nنوع Blood_Pressure قبل الإصلاح:', df['Blood_Pressure'].dtype)
print('أمثلة خام:', df['Blood_Pressure'].head(6).tolist())

# لماذا extract ثم to_numeric؟ to_numeric وحدها تفشل على " 130 mmHg".
# نستخرج الرقم أولاً بتعبير نمطي، ثم نحوّل. كلمة Unknown لا تحتوي رقماً فتصبح NaN.
bp_digits = df['Blood_Pressure'].astype(str).str.extract(r'(\d+\.?\d*)')[0]
df['Blood_Pressure'] = pd.to_numeric(bp_digits, errors='coerce')

print('نوع Blood_Pressure بعد الإصلاح:', df['Blood_Pressure'].dtype)
print('أمثلة بعد الإصلاح:', df['Blood_Pressure'].head(6).tolist())

# ---- قاعدة منطقية للقيم المستحيلة ----
# عمر -5 أو 190 ليس "شاذاً إحصائياً" بل خطأ إدخال. الأفضل تحويله إلى NaN ثم تعويضه،
# لأننا لو تركناه لتلوّث المتوسط والوسيط والحدود الربيعية.
impossible = (df['Age'] <= 0) | (df['Age'] > 120)
print(f'\nأعمار مستحيلة حُوّلت إلى NaN: {impossible.sum()}')
df.loc[impossible, 'Age'] = np.nan

print('\nالأنواع الآن:')
print(df.dtypes)
```

```python
# ==============================================================
# الخلية 5: تعويض القيم المفقودة (Imputation) بذكاء
# ==============================================================
print('الفراغات قبل المعالجة:')
print(df[['Age', 'Blood_Pressure', 'Cholesterol', 'Smoking_Status']].isna().sum())

# ---- 1) العمود الفئوي: المنوال ----
# لا معنى لمتوسط كلمتي "Never" و"Current"؛ الخيار المنطقي هو الأكثر تكراراً.
# مهم: نفعل هذا أولاً لأن خطوة التعويض الذكية أدناه تعتمد على Smoking_Status.
mode_smoking = df['Smoking_Status'].mode()[0]
df['Smoking_Status'] = df['Smoking_Status'].fillna(mode_smoking)
print('\nالمنوال المستخدم لحالة التدخين:', mode_smoking)

# ---- 2) العمر: وسيط عام ----
# الوسيط وليس المتوسط، لأن بعض الأعمار الشاذة (105-112) قد تسحب المتوسط للأعلى.
df['Age'] = df['Age'].fillna(df['Age'].median())

# ---- 3) لماذا الوسيط "لكل فئة" أذكى من الوسيط العام؟ ----
# نظرة سريعة: هل يختلف الكوليسترول فعلاً بين فئات التدخين؟
compare = df.groupby('Smoking_Status')[['Cholesterol', 'Blood_Pressure']].median()
compare.loc['(الوسيط العام)'] = df[['Cholesterol', 'Blood_Pressure']].median()
print('\nالوسيط حسب فئة التدخين مقابل الوسيط العام:')
display(compare)
# إذا كان المدخن الحالي يملك وسيطاً أعلى، فتعويض فراغه بالوسيط العام يجعله يبدو "أصح" مما هو عليه،
# أي نُدخل انحيازاً صامتاً في البيانات. التعويض داخل المجموعة يحافظ على هذا الفرق الحقيقي.

for col in ['Cholesterol', 'Blood_Pressure']:
    group_median = df.groupby('Smoking_Status')[col].transform('median')
    df[col] = df[col].fillna(group_median)

print('\nالفراغات بعد المعالجة:')
print(df[['Age', 'Blood_Pressure', 'Cholesterol', 'Smoking_Status']].isna().sum())
assert df[['Age', 'Blood_Pressure', 'Cholesterol', 'Smoking_Status']].isna().sum().sum() == 0
```

  <div dir="ltr" style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; background:#141414; border-radius:8px; padding:8px 12px; color:#FFFFFF; text-align:center;">


$$\text{IQR} = Q_3 - Q_1$$

$$\text{Lower Bound} = Q_1 - 1.5 \times \text{IQR}, \quad \text{Upper Bound} = Q_3 + 1.5 \times \text{IQR}$$

</div>
  <div style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; display: flex; gap: 14px; flex-wrap: wrap; margin-top: 16px;">
    <div style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; flex: 1; min-width:min(260px,100%); background: #141414; padding: 14px; border-radius: 8px; border-top: 3px solid #737373;">

      <ul style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; color: #D1D5DB; font-size: 13px; line-height: 1.7; padding-right: 18px; margin: 8px 0 0 0;">
        <li>نحذف الصف بالكامل.</li>
        <li>مناسب عندما تكون الشواذ قليلة جداً ونثق أنها خطأ.</li>
        <li><strong style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; color:#DC2626;">العيب:</strong> نخسر بقية معلومات المريض (وزنه، مدينته...).</li>
      </ul>

      <div style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; font-weight: bold; color: #DC2626;">التقليم (Capping / Winsorizing)</div>
      <ul style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; color: #D1D5DB; font-size: 13px; line-height: 1.7; padding-right: 18px; margin: 8px 0 0 0;">
        <li>نُبقي الصف ونقصّ القيمة إلى أقرب حد (Lower/Upper).</li>
        <li>مناسب عندما تكون البيانات قليلة ولا نريد خسارة أي سجل.</li>
        <li><strong style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; color:#DC2626;">العيب:</strong> قد يخفي حالة حقيقية نادرة، فنراجعها قبل القص.</li>
      </ul>

```python
# ==============================================================
# الخلية 6: كشف وتقليم القيم الشاذة عبر IQR + مخطط صندوقي قبل/بعد
# ==============================================================

def cap_outliers_iqr(data, column, k=1.5):
    """تقليم (Capping) القيم الشاذة في عمود رقمي بطريقة IQR."""
    q1, q3 = data[column].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - k * iqr, q3 + k * iqr
    n_out = ((data[column] < lower) | (data[column] > upper)).sum()
    # clip بدل الحذف: نحتفظ بالصف ونقصّ القيمة، فلا نخسر بقية معلومات المريض
    data[column] = data[column].clip(lower, upper)
    print(f'{column:15s} | الحدان: [{lower:7.1f}, {upper:7.1f}] | قيم شاذة قُلّمت: {n_out}')
    return data

cols_to_fix = ['Age', 'Blood_Pressure', 'Cholesterol']
df_before = df.copy()   # نحتفظ بنسخة لنقارن بها، وإلا ضاعت الحالة "قبل"

for c in cols_to_fix:
    df = cap_outliers_iqr(df, c)

# ---- الرسم بألوان الهوية: أسود + أحمر + أبيض ----
fig, axes = plt.subplots(2, 3, figsize=(13, 6), facecolor='#0B0B0B')
for j, c in enumerate(cols_to_fix):
    for i, (data, label) in enumerate([(df_before, 'Before'), (df, 'After')]):
        ax = axes[i, j]
        ax.set_facecolor('#141414')
        ax.boxplot(data[c], vert=False, patch_artist=True, widths=0.5,
                   boxprops=dict(facecolor='#991B1B', edgecolor='#FFFFFF'),
                   medianprops=dict(color='#FFFFFF', linewidth=2),
                   whiskerprops=dict(color='#E5E7EB'), capprops=dict(color='#E5E7EB'),
                   flierprops=dict(marker='o', markerfacecolor='#DC2626', markeredgecolor='#DC2626', markersize=5))
        ax.set_title(f'{c} - {label}', color='#FFFFFF', fontsize=11)
        ax.tick_params(colors='#E5E7EB')
        ax.set_yticks([])
        for s in ax.spines.values():
            s.set_color('#262626')
plt.tight_layout()
plt.show()
```

```python
# ==============================================================
# الخلية 7: اشتقاق الميزات (Feature Creation)
# ==============================================================

# BMI: ندمج عمودين في ميزة تحمل معنى طبياً مباشراً (الطول يُحوَّل من سم إلى متر)
df['BMI'] = (df['Weight_kg'] / (df['Height_cm'] / 100) ** 2).round(1)

# نسبة النبض إلى الكوليسترول: مؤشر مركب بسيط.
# لماذا النسبة؟ النموذج الخطي لا يستطيع "اختراع" النسبة وحده من العمودين المنفصلين.
df['Pulse_Cholesterol_Ratio'] = (df['Pulse'] / df['Cholesterol']).round(3)

# تصنيف العمر عبر pd.cut: بعض الأنماط (مثل خطورة ما بعد الستين) تصبح أوضح بالفئات
df['Age_Group'] = pd.cut(df['Age'], bins=[0, 30, 45, 60, 120],
                         labels=['Under_30', '30_45', '45_60', 'Over_60'])

print('الميزات الجديدة:')
display(df[['Age', 'Age_Group', 'Weight_kg', 'Height_cm', 'BMI', 'Pulse_Cholesterol_Ratio']].head(8))
print('توزيع الفئات العمرية:')
print(df['Age_Group'].value_counts().sort_index())

# هل الميزة الجديدة مفيدة؟ نقارن متوسط الخطورة بين الفئات
print('\nمتوسط Heart_Risk لكل فئة عمرية:')
print(df.groupby('Age_Group', observed=True)['Heart_Risk'].mean().round(2))
```

```python
# ==============================================================
# الخلية 8: الترميز والتصدير إلى NumPy
# ==============================================================

# ---- 1) ترميز ترتيبي لحالة التدخين ----
# لماذا map يدوي؟ لأنه يضمن أن الترتيب Never < Former < Current كما نريد نحن،
# بينما الترميز التلقائي يرتّب أبجدياً (Current قبل Former!) فيفسد المعنى.
df['Smoking_Status'] = df['Smoking_Status'].map({'Never': 0, 'Former': 1, 'Current': 2})

# ---- 2) One-Hot للمدينة والفئة العمرية ----
# نحذف Patient_ID أولاً: هو معرّف وليس خاصية، ولو دخل النموذج لتعلّم الحفظ لا التعميم
df_model = df.drop(columns=['Patient_ID'])
df_clean = pd.get_dummies(df_model, columns=['City', 'Age_Group'], drop_first=True, dtype=int)

print('الأعمدة النهائية:', list(df_clean.columns))

# ---- 3) فحوص الجودة قبل التسليم لليوم الثالث ----
assert df_clean.isna().sum().sum() == 0, 'ما زالت هناك قيم مفقودة!'
assert df_clean.select_dtypes(exclude='number').shape[1] == 0, 'ما زال هناك عمود غير رقمي!'

# ---- 4) الفصل إلى X و y وتحويلهما إلى مصفوفات NumPy ----
X = df_clean.drop('Heart_Risk', axis=1).values
y = df_clean['Heart_Risk'].values

print('\nشكل X :', X.shape, '| النوع:', X.dtype)
print('شكل y :', y.shape, '| النوع:', y.dtype)
print('نسبة الفئة الإيجابية في y:', round(y.mean(), 3))

# نحفظ المصفوفات لنستخدمها مباشرة في اليوم الثالث دون إعادة التنظيف
np.save('X_day2.npy', X)
np.save('y_day2.npy', y)
print('\nتم الحفظ: X_day2.npy و y_day2.npy  ->  جاهزة للتدريب في اليوم الثالث')
```

```python
# ==============================================================
# الخلية 9: كود البداية للتحدي (Starter Code)
# ==============================================================
import numpy as np
import pandas as pd

rng_t = np.random.default_rng(7)
n = 300

tenure = rng_t.integers(1, 72, n).astype(float)
contract_clean = rng_t.choice(['Month-To-Month', 'One Year', 'Two Year'], n, p=[0.55, 0.25, 0.20])
internet_clean = rng_t.choice(['Dsl', 'Fiber', 'No'], n, p=[0.35, 0.45, 0.20])
base = pd.Series(internet_clean).map({'Dsl': 45, 'Fiber': 80, 'No': 22}).values
monthly_true = (base + rng_t.normal(0, 6, n)).round(2)

logit = (-0.4 - 0.04 * (tenure - 30) + 1.1 * (contract_clean == 'Month-To-Month')
         - 0.8 * (contract_clean == 'Two Year') + 0.02 * (monthly_true - 60))
churn = np.where(rng_t.random(n) < 1 / (1 + np.exp(-logit)), 'Yes', 'No')

# ---- نُفسد البيانات ----
tenure_col = tenure.copy()
tenure_col[rng_t.choice(n, 15, replace=False)] = np.nan
tenure_col[[3, 77]] = [-3, 999]                       # قيم مستحيلة

def messy_money(v):
    r = rng_t.random()
    if r < 0.06: return 'N/A'
    if r < 0.30: return f'${v}'
    if r < 0.45: return f' {v} '
    return str(v)
monthly_col = [messy_money(v) for v in monthly_true]
monthly_col[10], monthly_col[200] = '900', '1500'     # شواذ

def messy_text(s):
    r = rng_t.random()
    if r < 0.6: return s
    if r < 0.8: return s.lower() + ' '
    return ' ' + s.upper()

tel_raw = pd.DataFrame({
    'Customer_ID': np.arange(5001, 5001 + n),
    'Tenure_Months': tenure_col,
    'Monthly_Charges': monthly_col,
    'Contract': [messy_text(s) for s in contract_clean],
    'Internet_Service': [messy_text(s) for s in internet_clean],
    'Churn': churn,
})
tel_raw = pd.concat([tel_raw, tel_raw.sample(10, random_state=3)]).sample(frac=1, random_state=5).reset_index(drop=True)
tel_raw.to_csv('telecom_churn_dirty.csv', index=False)

df_tel = pd.read_csv('telecom_churn_dirty.csv')
print('الأبعاد:', df_tel.shape)
display(df_tel.head())
df_tel.info()

# ==================== مساحة عملك ====================
# TODO 1: احذف الصفوف المكررة
# TODO 2: وحّد نصوص Contract و Internet_Service (strip + title)
# TODO 3: حوّل Monthly_Charges إلى رقم (تلميح: احذف كل ما ليس رقماً أو نقطة ثم to_numeric)
# TODO 4: حوّل Tenure_Months المستحيل (<=0 أو >120) إلى NaN
# TODO 5: عوّض Tenure_Months بوسيط كل نوع عقد، و Monthly_Charges بوسيط كل خدمة إنترنت
# TODO 6: قلّم الشواذ في Monthly_Charges بطريقة IQR
# TODO 7: اشتق ميزة جديدة (مثلاً Tenure_Group عبر pd.cut)
# TODO 8: رمّز Contract ترتيبياً، و Internet_Service + Tenure_Group بـ get_dummies(drop_first=True)
# TODO 9: حوّل Churn إلى 0/1 وصدّر X و y
# TODO 10: تحقق أنه لا توجد NaN وأن كل الأعمدة رقمية (assert)
```

```python
# ==============================================================
# الخلية 10: الحل النموذجي الكامل + تحقق تلقائي
# ==============================================================
t = pd.read_csv('telecom_churn_dirty.csv')   # نبدأ من الملف الخام لنُثبت أن الحل مستقل

# 1) المكررات
t = t.drop_duplicates().reset_index(drop=True)
assert len(t) == 300, 'يجب أن يبقى 300 عميل فريد'

# 2) توحيد النصوص (يحوّل "month-to-month " و " MONTH-TO-MONTH" إلى شكل واحد)
for c in ['Contract', 'Internet_Service']:
    t[c] = t[c].str.strip().str.title()

# 3) المبالغ: نحذف أي رمز غير رقمي ($ ومسافات)، وN/A تصبح نصاً فارغاً ثم NaN
t['Monthly_Charges'] = pd.to_numeric(
    t['Monthly_Charges'].astype(str).str.replace(r'[^0-9.]', '', regex=True), errors='coerce')

# 4) قيم مستحيلة -> NaN (قاعدة منطقية قبل أي إحصاء)
t.loc[(t['Tenure_Months'] <= 0) | (t['Tenure_Months'] > 120), 'Tenure_Months'] = np.nan

# 5) التعويض بوسيط كل مجموعة (العقد يؤثر في المدة، والخدمة تؤثر في السعر)
t['Tenure_Months'] = t['Tenure_Months'].fillna(t.groupby('Contract')['Tenure_Months'].transform('median'))
t['Monthly_Charges'] = t['Monthly_Charges'].fillna(t.groupby('Internet_Service')['Monthly_Charges'].transform('median'))

# 6) تقليم الشواذ بـ IQR (قيم 900 و1500 غير منطقية لباقة شهرية)
q1, q3 = t['Monthly_Charges'].quantile([0.25, 0.75])
iqr = q3 - q1
t['Monthly_Charges'] = t['Monthly_Charges'].clip(q1 - 1.5 * iqr, q3 + 1.5 * iqr)

# 7) ميزة جديدة
t['Tenure_Group'] = pd.cut(t['Tenure_Months'], bins=[0, 12, 36, 120], labels=['New', 'Mid', 'Loyal'])

# 8) الترميز
t['Contract'] = t['Contract'].map({'Month-To-Month': 0, 'One Year': 1, 'Two Year': 2})
t['Churn'] = t['Churn'].str.strip().str.title().map({'No': 0, 'Yes': 1})
t = t.drop(columns=['Customer_ID'])
t_clean = pd.get_dummies(t, columns=['Internet_Service', 'Tenure_Group'], drop_first=True, dtype=int)

# 9) التصدير
X_tel = t_clean.drop('Churn', axis=1).values
y_tel = t_clean['Churn'].values

# 10) التحقق التلقائي
assert t_clean.isna().sum().sum() == 0, 'توجد قيم NaN'
assert t_clean.select_dtypes(exclude='number').shape[1] == 0, 'يوجد عمود غير رقمي'
assert X_tel.shape[0] == y_tel.shape[0] == 300
assert set(np.unique(y_tel)) <= {0, 1}
assert t_clean['Monthly_Charges'].max() < 400, 'الشواذ لم تُقلَّم'

print('نجحت جميع الاختبارات')
print('شكل X:', X_tel.shape, '| شكل y:', y_tel.shape)
print('الأعمدة:', list(t_clean.drop("Churn", axis=1).columns))
```
