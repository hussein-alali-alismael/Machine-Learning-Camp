```python
# ==============================================================
# المرحلة 1 - الخلية 1: توليد بيانات تسرب العملاء الفوضوية (1200 صف)
# ==============================================================
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score, cross_val_predict
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
                             confusion_matrix, ConfusionMatrixDisplay)
from matplotlib.colors import LinearSegmentedColormap

# ألوان الهوية البصرية: نعرّفها مرة واحدة لتتطابق كل رسوم المشروع
RED, BLACK, DGRAY, GRID = '#DC2626', '#0B0B0B', '#374151', '#E5E7EB'
plt.rcParams.update({
    'figure.facecolor': 'white', 'axes.facecolor': 'white',
    'axes.grid': True, 'grid.color': GRID, 'grid.alpha': 0.8,
    'axes.spines.top': False, 'axes.spines.right': False, 'font.size': 11,
})

rng = np.random.default_rng(42)
N_UNIQUE, N_DUP = 1175, 25          # 1175 عميلاً فريداً + 25 مكرراً = 1200 صف

# ---------- القيم النظيفة أولاً ----------
tenure = rng.integers(0, 72, N_UNIQUE).astype(float)
tenure[rng.random(N_UNIQUE) < 0.03] = 0                      # عملاء جدد لم يكمل أي منهم شهراً
contract = rng.choice(['Month-to-month', 'One year', 'Two year'], N_UNIQUE, p=[0.55, 0.25, 0.20])
payment = rng.choice(['Electronic check', 'Mailed check', 'Bank transfer', 'Credit card'], N_UNIQUE, p=[0.35, 0.20, 0.22, 0.23])
monthly = (35 + rng.normal(0, 18, N_UNIQUE) + 10 * (payment == 'Electronic check')).clip(18, 120).round(2)
tickets = rng.poisson(0.8 + 1.4 * (contract == 'Month-to-month'))
total = (monthly * tenure * (1 + rng.normal(0, 0.03, N_UNIQUE))).round(2)

# التسرب يعتمد على عوامل منطقية: عقد شهري، قِصر المدة، كثرة الشكاوى، الفاتورة العالية، الدفع الإلكتروني
logit = (-1.9 + 1.5 * (contract == 'Month-to-month') - 1.0 * (contract == 'Two year')
         - 0.035 * (tenure - 30) + 0.5 * (tickets - 1.5) + 0.02 * (monthly - 65)
         + 0.6 * (payment == 'Electronic check'))
churn = (rng.random(N_UNIQUE) < 1 / (1 + np.exp(-logit))).astype(int)

# ---------- ثم نفسد البيانات عمداً ----------
monthly_col = monthly.copy()
monthly_col[rng.choice(N_UNIQUE, 59, replace=False)] = np.nan                       # فراغات 5%
monthly_col[rng.choice(N_UNIQUE, 3, replace=False)] = [950, 1100, 870]               # شواذ متطرفة
tickets_col = tickets.astype(float)
tickets_col[rng.choice(N_UNIQUE, 3, replace=False)] = [40, 55, 48]

def messy_contract(v):
    r = rng.random()
    if r < 0.04: return np.nan                         # فراغ
    if r < 0.16: return v.lower() + ' '                # حروف صغيرة ومسافة زائدة
    return v
contract_col = [messy_contract(v) for v in contract]

def messy_total(v, t):
    if t == 0: return 'New_Customer'                   # نص بدل الرقم
    r = rng.random()
    if r < 0.25: return f' ${v} '                      # رمز عملة ومسافات
    if r < 0.30: return f'${v:,.2f}'                   # فاصل آلاف
    return str(v)
total_col = [messy_total(v, t) for v, t in zip(total, tenure)]
payment_col = [p if rng.random() > 0.1 else p.upper() for p in payment]

raw = pd.DataFrame({
    'Customer_ID': [f'C{i:05d}' for i in range(1, N_UNIQUE + 1)],
    'Tenure_Months': tenure, 'Monthly_Charges': monthly_col, 'Total_Charges': total_col,
    'Contract_Type': contract_col, 'Support_Tickets': tickets_col,
    'Payment_Method': payment_col, 'Churn': churn,
})
# 25 صفاً مكرراً بالكامل ثم خلط الترتيب (المكررات في الواقع لا تأتي متتالية)
raw = pd.concat([raw, raw.sample(N_DUP, random_state=7)]).sample(frac=1, random_state=1).reset_index(drop=True)
raw.to_csv('raw_telecom_churn.csv', index=False)

# نقرأ من الملف تماماً كما يحدث مع بيانات حقيقية، ونحتفظ بنسخة احتياطية قبل أي جراحة
df = pd.read_csv('raw_telecom_churn.csv')
df_raw = df.copy()
print('تم إنشاء raw_telecom_churn.csv | الأبعاد:', df.shape)
```

```python
# ==============================================================
# المرحلة 2 - الخلية 3: جراحة تنظيف البيانات
# ==============================================================
df = df_raw.copy()                        # نبدأ دائماً من النسخة الاحتياطية الآمنة

# 1) المكررات أولاً: كل إحصاء لاحق (وسيط، IQR) يتأثر بها
n0 = len(df)
df = df.drop_duplicates().reset_index(drop=True)
print(f'حُذف {n0 - len(df)} مكرراً -> {len(df)} عميل فريد')
assert len(df) == 1175

# 2) توحيد النصوص (حروف صغيرة ومسافات)
df['Contract_Type'] = df['Contract_Type'].str.strip().str.capitalize()    # 'month-to-month ' -> 'Month-to-month'
df['Payment_Method'] = df['Payment_Method'].str.strip().str.capitalize()   # 'ELECTRONIC CHECK' -> 'Electronic check'

# 3) Total_Charges: نحذف كل ما ليس رقماً أو نقطة، و"New_Customer" تتحول إلى نص فارغ ثم NaN
df['Total_Charges'] = pd.to_numeric(
    df['Total_Charges'].astype(str).str.replace(r'[^0-9.]', '', regex=True), errors='coerce')

# 4) التعويض: الفئوي بالمنوال أولاً لأن تعويض الأرقام التالي يعتمد عليه
df['Contract_Type'] = df['Contract_Type'].fillna(df['Contract_Type'].mode()[0])
# الفاتورة الشهرية: وسيط نوع العقد نفسه (العقود تختلف في الأسعار) أدق من وسيط عام
df['Monthly_Charges'] = df['Monthly_Charges'].fillna(df.groupby('Contract_Type')['Monthly_Charges'].transform('median'))
# إجمالي المدفوع للعميل الجديد منطقياً = الفاتورة الشهرية × مدة الاشتراك (= 0 لمن لم يكمل شهراً)
df['Total_Charges'] = df['Total_Charges'].fillna(df['Monthly_Charges'] * df['Tenure_Months'])

# 5) تقليم الشواذ بـ IQR (Capping): نحافظ على الصف ونقصّ القيمة المتطرفة
def cap_iqr(data, col, k=1.5):
    q1, q3 = data[col].quantile([0.25, 0.75])
    lo, hi = q1 - k * (q3 - q1), q3 + k * (q3 - q1)
    n_out = int(((data[col] < lo) | (data[col] > hi)).sum())
    data[col] = data[col].clip(lo, hi)
    print(f'{col:16s} الحدان [{lo:8.1f}, {hi:8.1f}] | قُلّم: {n_out}')

for col in ['Monthly_Charges', 'Support_Tickets', 'Total_Charges']:
    cap_iqr(df, col)

print('\nالفراغات المتبقية:', int(df.isna().sum().sum()))
assert df.isna().sum().sum() == 0
display(df.describe().round(1))
```

```python
# ==============================================================
# المرحلة 3 - الخلية 4: هندسة الخصائص والترميز
# ==============================================================
# ---- الفخ: القسمة المباشرة على Tenure_Months ----
with np.errstate(divide='ignore', invalid='ignore'):
    naive = df['Support_Tickets'] / df['Tenure_Months']
print('قسمة مباشرة: inf =', int(np.isinf(naive).sum()), '| NaN =', int(naive.isna().sum()), '<- ستكسر النموذج')

# ---- الحل: +1 في المقام (عميل جديد = شهر واحد على الأقل) ----
df['Ticket_Rate'] = df['Support_Tickets'] / (df['Tenure_Months'] + 1)
# +1 الداخلية: متوسط الفاتورة التاريخي؛ +1 الخارجية: حماية من متوسط صفري
df['Charge_Spike_Ratio'] = df['Monthly_Charges'] / ((df['Total_Charges'] / (df['Tenure_Months'] + 1)) + 1)
assert np.isfinite(df[['Ticket_Rate', 'Charge_Spike_Ratio']]).all().all()

# ---- الترميز ----
# ترتيبي يدوي: الأبجدية ستخلط الترتيب (One year < Two year < Month...)
df['Contract_Type'] = df['Contract_Type'].map({'Month-to-month': 0, 'One year': 1, 'Two year': 2})
df_model = pd.get_dummies(df.drop(columns=['Customer_ID']), columns=['Payment_Method'], drop_first=True, dtype=int)

print('\nأعمدة الجدول النهائي:', list(df_model.columns))
display(df_model.head())
```

```python
# ==============================================================
# المرحلة 4 - الخلية 5: التقسيم الآمن + التطبيع + التجميع بدون تسرب
# ==============================================================
X = df_model.drop(columns=['Churn'])
y = df_model['Churn']
base_features = list(X.columns)

# 1) التقسيم أولاً مع stratify لأن التسرب فئة أقلية
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
print(f'تدريب: {X_train.shape} | اختبار: {X_test.shape}')
print(f'نسبة التسرب -> تدريب: {y_train.mean():.3f} | اختبار: {y_test.mean():.3f}')

# 2) التطبيع: fit على التدريب فقط
scaler = StandardScaler()
X_train_s = pd.DataFrame(scaler.fit_transform(X_train), columns=base_features, index=X_train.index)
X_test_s = pd.DataFrame(scaler.transform(X_test), columns=base_features, index=X_test.index)

# 3) التجميع: fit على التدريب، و predict فقط على الاختبار (الاختبار لا يشارك في تحديد المراكز)
kmeans = KMeans(n_clusters=3, n_init=10, random_state=42).fit(X_train_s)
seg_train = kmeans.predict(X_train_s)
seg_test = kmeans.predict(X_test_s)

# رقم الشريحة اسمي بلا ترتيب، فنحوله إلى أعمدة One-Hot (ونحذف الشريحة 0 كمرجع)
def add_segments(X_scaled, seg):
    out = X_scaled.copy()
    out['Segment_1'] = (seg == 1).astype(int)
    out['Segment_2'] = (seg == 2).astype(int)
    return out
X_train_f = add_segments(X_train_s, seg_train)
X_test_f = add_segments(X_test_s, seg_test)
feature_names = list(X_train_f.columns)

# ---- قراءة الشرائح تجارياً (على بيانات التدريب بوحداتها الأصلية) ----
profile = X_train[['Tenure_Months', 'Monthly_Charges', 'Support_Tickets']].copy()
profile['Segment'] = seg_train
profile['Churn'] = y_train
display(profile.groupby('Segment').agg(Customers=('Churn', 'size'), Avg_Tenure=('Tenure_Months', 'mean'),
        Avg_Monthly=('Monthly_Charges', 'mean'), Avg_Tickets=('Support_Tickets', 'mean'),
        Churn_Rate=('Churn', 'mean')).round(2))
```

```python
# ==============================================================
# المرحلة 4 - الخلية 6: حلبة منافسة النماذج الثلاثة
# ==============================================================
arena_models = {
    'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
    'Decision Tree': DecisionTreeClassifier(max_depth=5, min_samples_leaf=10, random_state=42),   # العمق محدود لمنع الحفظ
    'Random Forest': RandomForestClassifier(n_estimators=200, min_samples_leaf=3, random_state=42),
}
# shuffle=True لأن الصفوف قد تكون مرتبة؛ stratified للحفاظ على نسبة التسرب في كل جزء
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

rows = []
for name, model in arena_models.items():
    # التحقق المتقاطع على التدريب فقط: الاختبار يبقى مخبأً حتى النهاية
    cv_f1 = cross_val_score(model, X_train_f, y_train, cv=cv, scoring='f1').mean()
    model.fit(X_train_f, y_train)
    pred = model.predict(X_test_f)
    proba = model.predict_proba(X_test_f)[:, 1]
    rows.append({'Model': name, 'CV_F1': cv_f1,
                 'Accuracy': accuracy_score(y_test, pred), 'Precision': precision_score(y_test, pred, zero_division=0),
                 'Recall': recall_score(y_test, pred), 'F1_Score': f1_score(y_test, pred), 'ROC_AUC': roc_auc_score(y_test, proba)})

# نرتب بحسب CV_F1 (وليس بدرجة الاختبار) كي لا نختار الفائز بعد "رؤية" الاختبار
results = pd.DataFrame(rows).sort_values('CV_F1', ascending=False).set_index('Model').round(3)
display(results)

best_name = results.index[0]
best_model = arena_models[best_name]
print(f'الفائز بحسب التحقق المتقاطع: {best_name}')
```

<div dir="rtl" style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; background-color: #141414; color: #FFFFFF; padding: 20px; border-radius: 10px; border: 1px solid #262626; border-right: 5px solid #DC2626; font-family: 'Segoe UI', Tahoma, sans-serif; margin-bottom: 16px;"><h2 style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; color: #FFFFFF; margin-top: 0; font-size: 20px; border-bottom: 1px solid #262626; padding-bottom: 8px;"><span style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; color: #DC2626;">04 |</span> لوحة القيادة التنفيذية (Executive Visual Dashboard) وضبط العتبة</h2><p style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; color: #D1D5DB; font-size: 14.5px; line-height: 1.7; margin: 8px 0;">بعد اختيار الفائز، نجمع النتائج في لوحة واحدة يفهمها المدير في دقيقة. ونخفض عتبة القرار من 0.50 إلى

. ركّز على الخانة السفلية اليسرى: FN أي متسرب فاتنا.</div></div><div style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; flex: 1; min-width:min(240px,100%); background: #0B0B0B; padding: 14px; border-radius: 8px; border-top: 3px solid #DC2626;"><div style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; color: #DC2626; font-weight: bold; margin-bottom: 6px;">اللوحة 3 (أسفل اليسار): منحنى العتبات</div><div style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; color: #D1D5DB; font-size: 13.5px; line-height: 1.7;">كيف يتغير Precision و Recall و F1 مع العتبة. الخط الأحمر العمودي هو قرارنا.</div></div><div style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; flex: 1; min-width:min(240px,100%); background: #0B0B0B; padding: 14px; border-radius: 8px; border-top: 3px solid #FFFFFF;"><div style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; color: #FFFFFF; font-weight: bold; margin-bottom: 6px;">اللوحة 4 (أسفل اليمين): أهم الميزات</div><div style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; color: #D1D5DB; font-size: 13.5px; line-height: 1.7;">أكثر 8 ميزات تأثيراً في قرار النموذج: تخبرنا لماذا يعتقد أن العميل سيغادر.</div></div></div><p style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; color: #D1D5DB; font-size: 14.5px; line-height: 1.7; margin: 8px 0;"><strong style="box-sizing:border-box; max-width:100%; overflow-wrap:anywhere; color:#FFFFFF;">المقايضة:</strong> خفض

$\tau$

يرفع Recall (نصطاد أكثر) لكنه يزيد الإنذارات الكاذبة FP، فينخفض Precision. المهم أن نختار نقطة نقبلها تجارياً.</p></div>

```python
# ==============================================================
# المرحلة 5 - الخلية 7: لوحة القيادة الرباعية 2x2 وضبط العتبة
# ==============================================================
OPTIMAL_THRESHOLD = 0.35        # أقل من 0.5 لأن تفويت عميل متسرب أغلى من مكالمة احتفاظ إضافية
red_cmap = LinearSegmentedColormap.from_list('brand', ['#FFFFFF', RED])

fig, axes = plt.subplots(2, 2, figsize=(15, 11))
fig.suptitle('Customer Churn - Executive Dashboard', fontsize=17, fontweight='bold', color=BLACK)

# ---- [0,0] مقارنة النماذج الثلاثة عند العتبة المختارة ----
names = list(arena_models)
rec_vals, f1_vals = [], []
for name in names:
    p = arena_models[name].predict_proba(X_test_f)[:, 1]
    pred = (p >= OPTIMAL_THRESHOLD).astype(int)
    rec_vals.append(recall_score(y_test, pred))
    f1_vals.append(f1_score(y_test, pred))
pos = np.arange(len(names)); w = 0.38
ax = axes[0, 0]
ax.bar(pos - w / 2, rec_vals, w, color=RED, label='Recall')
ax.bar(pos + w / 2, f1_vals, w, color=BLACK, label='F1-Score')
for i in range(len(names)):
    ax.text(pos[i] - w / 2, rec_vals[i] + 0.01, f'{rec_vals[i]:.2f}', ha='center', fontsize=10)
    ax.text(pos[i] + w / 2, f1_vals[i] + 0.01, f'{f1_vals[i]:.2f}', ha='center', fontsize=10)
ax.set_xticks(pos); ax.set_xticklabels(names)
ax.set_ylim(0, 1.05)
ax.set_title(f'Model comparison at tau = {OPTIMAL_THRESHOLD}')
ax.legend()

# ---- [0,1] مصفوفة الارتباك للفائز ----
proba_best = best_model.predict_proba(X_test_f)[:, 1]
pred_best = (proba_best >= OPTIMAL_THRESHOLD).astype(int)
ConfusionMatrixDisplay.from_predictions(y_test, pred_best, ax=axes[0, 1], cmap=red_cmap, colorbar=False,
                                        display_labels=['Stayed', 'Churned'])
axes[0, 1].grid(False)
axes[0, 1].set_title(f'{best_name}: confusion matrix (tau = {OPTIMAL_THRESHOLD})')

# ---- [1,0] منحنى المقايضة عبر العتبات 0.1 .. 0.9 ----
taus = np.round(np.arange(0.10, 0.91, 0.05), 2)
P = [precision_score(y_test, (proba_best >= t).astype(int), zero_division=0) for t in taus]
R = [recall_score(y_test, (proba_best >= t).astype(int)) for t in taus]
F = [f1_score(y_test, (proba_best >= t).astype(int), zero_division=0) for t in taus]
ax = axes[1, 0]
ax.plot(taus, R, color=RED, linewidth=2.5, label='Recall')
ax.plot(taus, P, color=BLACK, linewidth=2.5, label='Precision')
ax.plot(taus, F, color=DGRAY, linewidth=2, linestyle='--', label='F1')
ax.axvline(OPTIMAL_THRESHOLD, color=RED, linewidth=2, linestyle=':')
ax.text(OPTIMAL_THRESHOLD + 0.01, 0.05, f'tau = {OPTIMAL_THRESHOLD}', color=RED, fontweight='bold')
ax.set_xlabel('Decision threshold'); ax.set_ylabel('Score'); ax.set_ylim(0, 1.05)
ax.set_title('Threshold trade-off'); ax.legend()

# ---- [1,1] أهم 8 ميزات ----
if hasattr(best_model, 'feature_importances_'):
    imp = pd.Series(best_model.feature_importances_, index=feature_names)
else:
    imp = pd.Series(np.abs(best_model.coef_[0]), index=feature_names)   # الخطي: حجم الوزن بعد التطبيع
top8 = imp.sort_values().tail(8)
ax = axes[1, 1]
ax.barh(top8.index, top8.values, color=[DGRAY] * 7 + [RED])           # الأهم باللون الأحمر
ax.set_title('Top 8 features'); ax.set_xlabel('Importance')

plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()

rec_default = recall_score(y_test, (proba_best >= 0.5).astype(int))
rec_tuned = recall_score(y_test, pred_best)
print(f'Recall للفائز: عند 0.50 = {rec_default:.3f}  ->  عند {OPTIMAL_THRESHOLD} = {rec_tuned:.3f}')
print('أهم 3 ميزات:', list(top8.index[::-1][:3]))
```

```python
# ==============================================================
# المرحلة 6 - الخلية 8: حفظ النظام ودالة التوقع الحي
# ==============================================================
pipeline_bundle = {
    'model': best_model, 'scaler': scaler, 'kmeans': kmeans,
    'base_features': base_features, 'feature_names': feature_names,
    'optimal_threshold': OPTIMAL_THRESHOLD, 'model_name': best_name,
}
joblib.dump(pipeline_bundle, 'churn_prediction_system.pkl')
print('تم الحفظ:', 'churn_prediction_system.pkl', f'({os.path.getsize("churn_prediction_system.pkl") / 1024:.0f} KB)')

def predict_customer_risk(customer_dict, bundle_path='churn_prediction_system.pkl'):
    """تستقبل بيانات عميل خام وتُرجع تقرير المخاطر. تكرر نفس مراحل التدريب بالترتيب نفسه."""
    b = joblib.load(bundle_path)
    new = pd.DataFrame([customer_dict])

    # 1) نفس هندسة الخصائص (+1 تحمي من القسمة على صفر)
    new['Ticket_Rate'] = new['Support_Tickets'] / (new['Tenure_Months'] + 1)
    new['Charge_Spike_Ratio'] = new['Monthly_Charges'] / ((new['Total_Charges'] / (new['Tenure_Months'] + 1)) + 1)
    new['Contract_Type'] = new['Contract_Type'].map({'Month-to-month': 0, 'One year': 1, 'Two year': 2})

    # 2) الترميز: get_dummies بلا drop_first (عينة واحدة ستفقد فئتها)، ثم reindex يحذف المرجع ويملأ الناقص بصفر
    new = pd.get_dummies(new, columns=['Payment_Method'], dtype=int)
    new = new.reindex(columns=b['base_features'], fill_value=0)

    # 3) التطبيع بإحصاءات التدريب المحفوظة (وليس إحصاءات هذه العينة!)
    scaled = pd.DataFrame(b['scaler'].transform(new), columns=b['base_features'])

    # 4) الشريحة السلوكية ثم الأعمدة النهائية بالترتيب المحفوظ
    seg = int(b['kmeans'].predict(scaled)[0])
    scaled['Segment_1'] = int(seg == 1)
    scaled['Segment_2'] = int(seg == 2)
    scaled = scaled.reindex(columns=b['feature_names'], fill_value=0)

    # 5) التوقع بالعتبة المحفوظة
    prob = float(b['model'].predict_proba(scaled)[0, 1])
    decision = 'HIGH RISK' if prob >= b['optimal_threshold'] else 'STABLE'
    print('-' * 52)
    print(f"  Probability of churn : {prob:6.1%}")
    print(f"  Threshold            : {b['optimal_threshold']:.2f}")
    print(f"  Segment              : {seg}")
    print(f"  Decision             : {decision}")
    print('-' * 52)
    return {'probability': prob, 'segment': seg, 'decision': decision}

print('\nعميل 1: جديد، عقد شهري، شكاوى كثيرة، فاتورة مرتفعة، دفع إلكتروني')
risky = predict_customer_risk({'Tenure_Months': 2, 'Monthly_Charges': 105.0, 'Total_Charges': 210.0,
                               'Contract_Type': 'Month-to-month', 'Support_Tickets': 5, 'Payment_Method': 'Electronic check'})
print('\nعميل 2: قديم، عقد سنتين، بلا شكاوى، فاتورة معتدلة، بطاقة ائتمان')
stable = predict_customer_risk({'Tenure_Months': 58, 'Monthly_Charges': 52.0, 'Total_Charges': 3016.0,
                                'Contract_Type': 'Two year', 'Support_Tickets': 0, 'Payment_Method': 'Credit card'})
assert risky['probability'] > stable['probability'], 'العميل الخطر يجب أن يحصل على احتمال أعلى'
```

```python
# ==============================================================
# التحدي - الخلية 10: الحل الكامل + اختبارات التخرج + شهادة الإتمام
# ==============================================================
# ---------- 1) EDA سريع ----------
print('نسبة المصابين:', round(heart['Heart_Disease'].mean(), 3))
fig, axes = plt.subplots(1, 2, figsize=(11, 3.8))
vc = heart['Heart_Disease'].value_counts(normalize=True).sort_index() * 100
axes[0].bar(['Healthy', 'Disease'], vc.values, color=[BLACK, RED]); axes[0].set_title('Class balance (%)')
axes[1].hist(heart.loc[heart['Heart_Disease'] == 0, 'Age'], bins=20, alpha=0.75, color=BLACK, label='Healthy')
axes[1].hist(heart.loc[heart['Heart_Disease'] == 1, 'Age'], bins=20, alpha=0.75, color=RED, label='Disease')
axes[1].set_title('Age by diagnosis'); axes[1].legend()
plt.tight_layout(); plt.show()

# ---------- 2) التنظيف ----------
h = heart_backup.copy()
h = h.drop_duplicates().reset_index(drop=True)
assert len(h) == 900, 'يجب أن يبقى 900 مريض فريد'

h['Sex'] = h['Sex'].str.strip().str.upper().str[0]                     # ' f', 'Male', 'M' -> 'M' / 'F'
h['Max_Heart_Rate'] = pd.to_numeric(h['Max_Heart_Rate'], errors='coerce')   # 'Unknown' -> NaN
h.loc[h['Cholesterol'] == 0, 'Cholesterol'] = np.nan                   # كوليسترول صفر مستحيل
for col in ['Cholesterol', 'Resting_BP', 'Max_Heart_Rate']:
    h[col] = h[col].fillna(h.groupby('Sex')[col].transform('median'))   # وسيط كل جنس

def cap(data, col, k=1.5):
    q1, q3 = data[col].quantile([0.25, 0.75])
    data[col] = data[col].clip(q1 - k * (q3 - q1), q3 + k * (q3 - q1))
for col in ['Cholesterol', 'Resting_BP', 'Max_Heart_Rate', 'ST_Depression']:
    cap(h, col)
assert h.isna().sum().sum() == 0

# ---------- 3) الخصائص والترميز ----------
h['HR_Reserve'] = (220 - h['Age']) - h['Max_Heart_Rate']      # الاحتياطي القلبي: الفرق بين الأقصى المتوقع والفعلي
h['Sex'] = h['Sex'].map({'M': 1, 'F': 0})
h['Exercise_Angina'] = h['Exercise_Angina'].map({'No': 0, 'Yes': 1})
h = pd.get_dummies(h, columns=['Chest_Pain'], drop_first=True, dtype=int)

# ---------- 4) التقسيم الآمن + التطبيع + منافسة النماذج ----------
Xh, yh = h.drop(columns=['Heart_Disease']), h['Heart_Disease']
h_features = list(Xh.columns)
Xh_tr, Xh_te, yh_tr, yh_te = train_test_split(Xh, yh, test_size=0.2, stratify=yh, random_state=42)
h_scaler = StandardScaler()
Xh_tr_s = pd.DataFrame(h_scaler.fit_transform(Xh_tr), columns=h_features)
Xh_te_s = pd.DataFrame(h_scaler.transform(Xh_te), columns=h_features)

cv_h = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
h_models = {'Logistic Regression': LogisticRegression(max_iter=1000, random_state=42),
            'Random Forest': RandomForestClassifier(n_estimators=200, min_samples_leaf=3, random_state=42)}
h_scores = {n: cross_val_score(m, Xh_tr_s, yh_tr, cv=cv_h, scoring='f1').mean() for n, m in h_models.items()}
h_best_name = max(h_scores, key=h_scores.get)
h_best = h_models[h_best_name].fit(Xh_tr_s, yh_tr)
print('F1 بالتحقق المتقاطع:', {k: round(float(v), 3) for k, v in h_scores.items()}, '-> الفائز:', h_best_name)

# ---------- 5) ضبط العتبة من بيانات التدريب فقط ----------
# احتمالات "خارج الطية" (Out-of-Fold): كل مريض يُتوقع بنموذج لم يره، فلا نغشّ ولا نلمس الاختبار
oof = cross_val_predict(h_models[h_best_name], Xh_tr_s, yh_tr, cv=cv_h, method='predict_proba')[:, 1]
candidates = np.round(np.arange(0.05, 0.96, 0.01), 2)
ok = [t for t in candidates if recall_score(yh_tr, (oof >= t).astype(int)) >= 0.92]    # هامش أمان فوق 0.88
h_tau = float(max(ok))                                                                  # أكبر عتبة تحقق الهدف = أقل إنذارات كاذبة

h_proba = h_best.predict_proba(Xh_te_s)[:, 1]
h_pred = (h_proba >= h_tau).astype(int)
h_recall = recall_score(yh_te, h_pred)
h_precision = precision_score(yh_te, h_pred)
print(f'العتبة المختارة = {h_tau:.2f} | Recall (اختبار) = {h_recall:.3f} | Precision = {h_precision:.3f} | AUC = {roc_auc_score(yh_te, h_proba):.3f}')

# ---------- 6) الحفظ والتحميل ----------
heart_bundle = {'model': h_best, 'scaler': h_scaler, 'feature_names': h_features, 'optimal_threshold': h_tau}
joblib.dump(heart_bundle, 'heart_diagnosis_system.pkl')
loaded = joblib.load('heart_diagnosis_system.pkl')

# ---------- اختبارات التخرج الآلية ----------
assert os.path.exists('heart_diagnosis_system.pkl'), 'ملف النظام غير موجود'
assert set(loaded) == {'model', 'scaler', 'feature_names', 'optimal_threshold'}, 'الحزمة ناقصة'
assert Xh_tr.shape[0] == 720 and Xh_te.shape[0] == 180, 'التقسيم يجب أن يكون 80/20'
assert abs(yh_tr.mean() - yh_te.mean()) < 0.01, 'stratify لم يُطبَّق'
assert np.allclose(Xh_tr_s.mean(axis=0), 0, atol=1e-8) and not np.allclose(Xh_te_s.mean(axis=0), 0, atol=1e-8), 'تسرب في التطبيع'
assert h_recall >= 0.88, f'Recall = {h_recall:.3f} أقل من 0.88'
# السلامة الإنتاجية: الحزمة المحمّلة تعطي نفس توقعات النموذج الأصلي
assert np.allclose(loaded['model'].predict_proba(Xh_te_s)[:, 1], h_proba), 'الحزمة المحمّلة تخالف الأصل'

# ---------- شهادة الإتمام ----------
from datetime import date
G, R, W, X0 = '\033[92m', '\033[91m', '\033[97m', '\033[0m'
line = '=' * 64
print(f"""
{R}{line}{X0}
{W}              شهادة إتمام برمجية | Certificate of Completion{X0}
{R}{line}{X0}
{G}  ✔ تم اجتياز جميع اختبارات التخرج الآلية بنجاح{X0}
{W}  مشروع 1: نظام الإنذار المبكر لتسرب العملاء (Customer Churn){X0}
{W}  مشروع 2: نظام التشخيص المبدئي لأمراض القلب (Recall = {h_recall:.2f}){X0}

{W}  تؤكد هذه الشهادة إتمام الطالب بنجاح متطلبات
  معسكر تعلم الآلة التطبيقي من الصفر في مركز Domains:
  Python  |  NumPy  |  Pandas  |  Regression  |  Classification
  Clustering  |  Pipelines  |  Model Deployment{X0}

{W}  بإشراف: المهندس حسين العلي الإسماعيل{X0}
{W}  التاريخ: {date.today().isoformat()}{X0}
{R}{line}{X0}
""")
```
